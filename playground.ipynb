```python
size="gemma-3-4b-it"
q_df=pd.DataFrame()
save_as=f"200_q_{size}.csv"
h,m,l,ll,cu=0,0,0,0,0
i=0
while True:
    row=data_shuffled.iloc[i]
    if row["level"] == "high level of confidence" and h<40:
        h += 1
        q_df = pd.concat([q_df, row.to_frame().T], ignore_index=True)
    elif row["level"] == "moderate level of confidence" and m<40:
        m += 1
        q_df = pd.concat([q_df, row.to_frame().T], ignore_index=True)
    elif row["level"] == "low level of confidence" and l<40:
        l += 1
        q_df = pd.concat([q_df, row.to_frame().T], ignore_index=True)
    elif row["level"] == "lowest level of confidence" and ll<40:
        ll += 1
        q_df = pd.concat([q_df, row.to_frame().T], ignore_index=True)
    elif row["level"] == "complete uncertainty, reject to reply" and cu<40:
        cu += 1
        q_df = pd.concat([q_df, row.to_frame().T], ignore_index=True)
    if h>=40 and m>=40 and l>=40 and ll>=40 and cu>=40:
        break
    i+=1
    if i%100==0:
        print(f"Processed {i} rows...")
print(q_df.head())
len(q_df)
```

```python
q_df.to_csv(save_as, index=False)
```

```python
import re

def clean_text(text):
    # Remove leading numbering OR bullet
    text = re.sub(r'^\s*(\d+[\.\)\-]|[-•])\s*', '', text)
    
    # Normalize spaces (including trailing)
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```
