```python
import pandas as pd
res=pd.read_csv(r"c:\Users\user\Downloads\results_google_gemma-3-4b-it.csv", encoding='utf-8')
h,m,l,ll,cu=0,0,0,0,0
for r in range(len(res)):
    if res["confidences"][r] > 0.8:
        h += 1  
    elif res["confidences"][r] > 0.6:
        m += 1
    elif res["confidences"][r] > 0.4:
        l += 1
    elif res["confidences"][r] > 0.2:
        ll += 1
    else:
        cu += 1
        
print(f"High Confidence: {h} ({h/len(res)*100:.2f}%)")
print(f"Moderate Confidence: {m} ({m/len(res)*100:.2f}%)")
print(f"Low Confidence: {l} ({l/len(res)*100:.2f}%)")
print(f"Lowest Confidence: {ll} ({ll/len(res)*100:.2f}%)")
print(f"Completely Uncertain: {cu} ({cu/len(res)*100:.2f}%)")
```
